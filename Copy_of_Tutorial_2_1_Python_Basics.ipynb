## 📘 Bahagian 1: Variables dan Data Types

**Variable** ialah bekas (*container*) untuk menyimpan data.  
Python mempunyai beberapa jenis data (*data type*) utama:

| Data Type | Contoh | Keterangan |
|---|---|---|
| `int` | `x = 10` | Nombor bulat |
| `float` | `y = 3.14` | Nombor perpuluhan |
| `str` | `nama = "AI"` | Teks / String |
| `bool` | `flag = True` | Benar / Palsu |
| `list` | `data = [1,2,3]` | Senarai |
| `dict` | `info = {"a":1}` | Kunci-nilai |


````python
# TODO: Isikan nilai yang sesuai untuk setiap variable
nama_pelajar = "MIOR"       # Nama anda (str)
umur         = 23         # Umur anda (int)
gpa          = 3.78         # GPA anda, contoh: 3.75 (float)
aktif        = "True"             # True atau False (bool)

print('=' * 40)
print('  MAKLUMAT PELAJAR')
print('=' * 40)
print(f'Nama    : {nama_pelajar}')
print(f'Umur    : {umur} tahun')
print(f'GPA     : {gpa}')
print(f'Aktif   : {aktif}')
print(f'Jenis umur : {type(umur)}')
print(f'Jenis gpa  : {type(gpa)}')
````

## 📘 Bahagian 3: Conditional Statement — if / elif / else

Digunakan untuk membuat keputusan dalam program berdasarkan syarat tertentu.

```python
if syarat1:
    # laksanakan ini
elif syarat2:
    # laksanakan ini
else:
    # laksanakan ini
```


````python
def klasifikasi_model(nama_model, ketepatan):
    """
    Fungsi untuk menilai prestasi model AI.
    Parameter:
        nama_model (str)   : Nama model
        ketepatan  (float) : Nilai ketepatan 0-100
    Return:
        status (str)
    """
    # TODO: Lengkapkan syarat berikut
    if ketepatan >= 90:        # Sangat Baik jika >= 90
        status = "Sangat Baik ✅"
    elif ketepatan >= 70:      # Baik jika >= 70
        status = "Baik 👍"
    elif ketepatan >= 50:      # Boleh Diterima jika >= 50
        status = "Boleh Diterima ⚠️"
    else:
        status = "Perlu Dipertingkat ❌"

    print(f'Model     : {nama_model}')
    print(f'Ketepatan : {ketepatan}%')
    print(f'Status    : {status}')
    print('-' * 35)
    return status

# Uji fungsi
klasifikasi_model("YOLOv8", 94.5)
klasifikasi_model("ANN_Asas", 48.0)
````

## 📘 Bahagian 4: Loop — for

Loop `for` digunakan untuk **mengulang** blok kod untuk setiap item dalam senarai.

```python
for item in senarai:
    # lakukan sesuatu pada item
```


````python
senarai_ujian = [
    ("YOLOv8",    94.5),
    ("CNN_Mudah", 65.2),
    ("ANN_Asas",  48.0),
    ("ResNet50",  91.3),
]

print('=' * 35)
print('  LAPORAN PRESTASI MODEL AI')
print('=' * 35)

# TODO: Gunakan for loop untuk panggil klasifikasi_model()
#       bagi setiap model dalam senarai_ujian
for nama_model, ketepatan in senarai_ujian:
    klasifikasi_model(nama_model, ketepatan)
````

---
## 📘 Bahagian 5: Loop — while

Loop `while` mengulang blok kod **selagi syarat masih benar**.

```python
while syarat:
    # lakukan sesuatu
```


````python
# Simulasi proses latihan model
epoch     = 0
ketepatan = 50.0

# TODO: Lengkapkan syarat while — berhenti apabila ketepatan >= 95
while ketepatan < 95:
    epoch     += 1
    ketepatan += 5.0
    print(f'Epoch {epoch:02d} | Ketepatan: {ketepatan:.1f}%')

print(f'\nLatihan selesai pada Epoch {epoch}!')
print(f'Ketepatan akhir: {ketepatan:.1f}%')
````
